```python
import os
```

```python
# Source folder to scan. Examples:
#   DBFS:            "dbfs:/mnt/raw/incoming"
#   Unity Catalog:   "/Volumes/main/default/my_volume/incoming"
SOURCE_FOLDER = "/Volumes/20717-04/landing/raw_data/All_Data/"
 
# Destination root folder. Each file gets its own subfolder under here.
DEST_FOLDER = "/Volumes/20717-04/landing/raw_data"
 
# If True, subfolder name = filename WITHOUT extension (e.g. "report_jan")
# If False, subfolder name = full filename including extension
STRIP_EXTENSION_FOR_FOLDER_NAME = True
```

```python
def is_dbfs_path(path: str) -> bool:
    return path.startswith("dbfs:/")
 
def list_files(folder: str):
    """Return a list of (name, full_path) for files (not directories) in folder."""
    if is_dbfs_path(folder):
        entries = dbutils.fs.ls(folder)
        return [(e.name.rstrip("/"), e.path) for e in entries if not e.isDir()]
    else:
        entries = os.listdir(folder)
        results = []
        for name in entries:
            full_path = os.path.join(folder, name)
            if os.path.isfile(full_path):
                results.append((name, full_path))
        return results
 
def ensure_folder(path: str):
    if is_dbfs_path(path):
        dbutils.fs.mkdirs(path)
    else:
        os.makedirs(path, exist_ok=True)
 
def copy_file(src_path: str, dest_path: str):
    if is_dbfs_path(src_path):
        dbutils.fs.cp(src_path, dest_path)
    else:
        import shutil
        shutil.copy2(src_path, dest_path)
 
def folder_name_for(filename: str) -> str:
    if STRIP_EXTENSION_FOR_FOLDER_NAME:
        return os.path.splitext(filename)[0]
    return filename
```

```python
def main():
    files = list_files(SOURCE_FOLDER)
    if not files:
        print(f"No files found in {SOURCE_FOLDER}")
        return
 
    print(f"Found {len(files)} file(s) in {SOURCE_FOLDER}")
 
    for filename, src_path in files:
        subfolder_name = folder_name_for(filename)
        dest_folder_path = os.path.join(DEST_FOLDER, subfolder_name).replace("\\", "/")
        dest_file_path = os.path.join(dest_folder_path, filename).replace("\\", "/")
 
        ensure_folder(dest_folder_path)
        copy_file(src_path, dest_file_path)
 
        print(f"  {filename} -> {dest_file_path}")
 
    print("Done.")
```

```python
if __name__ == "__main__":
    main()
```
