```python
# Define your target path (DBFS, Unity Catalog Volumes, or mounted storage)
target_path = "/Volumes/20717-04/landing/raw_data"
srcArray  = []

# List all items and filter for folders
folders = [f.path for f in dbutils.fs.ls(target_path) if f.isDir()]

# Print the list of folders
for folder in folders:
    folderDetails = folder.split("/")
    #print(folderDetails[len(folderDetails)-2])
    if folderDetails[len(folderDetails)-2] != "All_Data":
        srcArray.append({"src" : folderDetails[len(folderDetails)-2]})

#print(srcArray) 
# DBTITLE 1
```

```python
dbutils.jobs.taskValues.set(key="output_key",value=srcArray)
```
