```python
dbutils.widgets.text("srcLanding","")
```

```python
srcFolder=dbutils.widgets.get("srcLanding")
```

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "parquet")\
    .option("cloudFiles.schemaLocation", f"/Volumes/20717-04/bronze/bronzevolume/{srcFolder}/checkpoint")\
    .option("cloudFiles.schemaEvolutionMode","rescue")\
    .load(f"/Volumes/20717-04/landing/raw_data/{srcFolder}/")
    
```

```python
df.writeStream.format("delta")\
    .outputMode("append")\
    .trigger(once=True)\
    .option("checkpointLocation", f"/Volumes/20717-04/bronze/bronzevolume/{srcFolder}/checkpoint")\
    .option("path",f"/Volumes/20717-04/bronze/bronzevolume/{srcFolder}")\
    .start()
    
```
